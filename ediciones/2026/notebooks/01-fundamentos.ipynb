## 2. Demostración modelada: El espectro de criticidad

En la sesión teórica aprendimos que la adopción de IA generativa no depende de la sofisticación técnica, sino del **costo del error** y la **reversibilidad del fallo**:

### Caso 1 · Resumen de minutas de comités comerciales (Bajo costo de error)
- **Valor de negocio:** Medio/Alto (ahorra ~45 minutos de digitación por reunión a cada jefatura comercial).
- **Costo del error:** Bajo (si el modelo omite un acuerdo menor, los asistentes lo subsanan de inmediato al revisar el correo).
- **Tolerancia al error:** Alta (falla inocua y fácilmente reversible).
- **Cuadrante asignado:** **Cuadrante A / C (Automatización asistida)**. El organizador da un vistazo rápido antes de enviar.

### Caso 2 · Dictamen vinculante de compliance y lavado de activos (Alto costo de error)
- **Valor de negocio:** Muy alto (cruce masivo de proveedores internacionales contra listas de sanciones).
- **Costo del error:** Catastrófico (sanciones penales para directores, multas regulatorias y pérdida de licencia bancaria).
- **Tolerancia al error:** Nula (falla irreversible con responsabilidad legal indelegable).
- **Cuadrante asignado:** **Cuadrante B (Co-piloto HITL obligatorio)**. La IA redacta el borrador de antecedentes, pero el oficial de cumplimiento audita y firma obligatoriamente la resolución.

## 3. Práctica guiada: El dilema de atención bancaria masiva

Una institución financiera en Chile recibe **15.000 solicitudes mensuales** en su canal web. La gerencia de operaciones propuso inicialmente:
*"Conectemos un LLM autónomo para responder el 100% de los mensajes de inmediato y eliminemos el tiempo de espera"*.

### Análisis de triaje y segmentación de flujos
En lugar de un enfoque "todo o nada", la arquitectura directiva correcta aplica un **triaje inteligente** que deriva cada solicitud según su criticidad:

| Tipo de Solicitud | Volumen Estimado | Tolerancia al Error | Cuadrante Matriz | Arquitectura de Supervisión |
|:---|:---:|:---:|:---:|:---|
| **Consultas informativas** (horarios, saldos, sucursales) | 8.000 / mes | Alta | **Cuadrante A** | Automatización asistida directa con respuestas precalibradas. |
| **Reclamos por cobros duplicados** | 4.500 / mes | Media | **Cuadrante A / B** | Extracción estructurada (IDP) de montos y cuentas; confirmación previa antes de abonar. |
| **Denuncias de fraude y bloqueos** | 1.800 / mes | Baja | **Cuadrante B** | Bloqueo preventivo automático + derivación inmediata a analista de seguridad. |
| **Solicitudes de crédito y renegociación** | 700 / mes | Nula | **Cuadrante B** | Evaluación asistida por IA; firma y visación final reservada exclusivamente a ejecutivo bancario. |

## 4. Simulador interactivo: Matriz Valor vs. Costo del Error

Utilicen la siguiente celda interactiva para evaluar cualquier proceso de su organización.

### Instrucciones:
1. Seleccionen un proceso preconfigurado o elijan **"Proceso propio de mi empresa"**.
2. Ajusten la capacidad requerida, el nivel de valor de negocio y el costo del error.
3. Hagan clic en el botón **Play** (▶) a la izquierda para generar el dictamen directivo y ver la ubicación en la matriz.

```python
#@title Matriz Interactiva de Priorización y Triaje de Procesos { run: "auto" }
#@markdown Ingrese o seleccione los parámetros del proceso organizacional a evaluar:

proceso_seleccionado = "Triaje y análisis de solicitudes bancarias (Caso Guiado)"  #@param ["Triaje y análisis de solicitudes bancarias (Caso Guiado)", "Minutas de comités comerciales (Caso 1)", "Dictamen vinculante de compliance (Caso 2)", "Proceso propio de mi empresa (Personalizado)"]
nombre_proceso_personalizado = "Conciliación de facturas de proveedores"  #@param {type:"string"}
capacidad_ia = "Extracción de datos estructurados (IDP)"  #@param ["Resumir textos extensos", "Clasificar y categorizar", "Extracción de datos estructurados (IDP)", "Reescribir y adaptar tono"]
nivel_valor = "Alto (Impacto estratégico o gran ahorro de horas)"  #@param ["Alto (Impacto estratégico o gran ahorro de horas)", "Medio (Ahorro operativo moderado)", "Bajo (Impacto marginal)"]
nivel_costo_error = "Alto (Pérdidas financieras severas, legal o regulatorio)"  #@param ["Crítico / Catastrófico (Sanciones penales o regulatorias)", "Alto (Pérdidas financieras severas, legal o regulatorio)", "Medio (Retrabajo interno subsanable)", "Bajo / Inocuo (Error evidente corregible de inmediato)"]
volumen_mensual = 15000  #@param {type:"integer"}

# Determinar nombre activo
if proceso_seleccionado == "Proceso propio de mi empresa (Personalizado)":
    nombre_activo = nombre_proceso_personalizado
else:
    nombre_activo = proceso_seleccionado.split(" (")[0]

# Mapeo a coordenadas cuantitativas en la matriz (0 a 10)
valor_map = {
    "Alto (Impacto estratégico o gran ahorro de horas)": 8.0,
    "Medio (Ahorro operativo moderado)": 5.0,
    "Bajo (Impacto marginal)": 2.5,
}
error_map = {
    "Crítico / Catastrófico (Sanciones penales o regulatorias)": 9.0,
    "Alto (Pérdidas financieras severas, legal o regulatorio)": 7.5,
    "Medio (Retrabajo interno subsanable)": 4.5,
    "Bajo / Inocuo (Error evidente corregible de inmediato)": 2.0,
}

y_val = valor_map.get(nivel_valor, 5.0)
x_err = error_map.get(nivel_costo_error, 5.0)

# Identificación de cuadrante
if y_val >= 5.0 and x_err < 5.0:
    cuadrante = "CUADRANTE A · Automatización Asistida"
    arquetipo = "Cíborg o Centauro liviano"
    supervision = "Revisión por muestreo o supervisión secundaria del usuario final."
    veredicto = "Adopción prioritaria. Entrega alto retorno con riesgo operativo controlado."
elif y_val >= 5.0 and x_err >= 5.0:
    cuadrante = "CUADRANTE B · Co-piloto con HITL Obligatorio"
    arquetipo = "Centauro estricto (Humano en el Bucle)"
    supervision = "Compuerta humana previa obligatoria. Ninguna salida tiene efecto legal o financiero sin firma profesional responsable."
    veredicto = "Adopción de alto valor pero condicionada a gobernanza estricta. Prohibida la automatización autónoma desatendida."
elif y_val < 5.0 and x_err < 5.0:
    cuadrante = "CUADRANTE C · Exploración y Eficiencia Interna"
    arquetipo = "Asistente individual libre"
    supervision = "Criterio del colaborador. No requiere comités burocráticos."
    veredicto = "Uso táctico para ganar agilidad individual. No priorizar inversión en desarrollo corporativo."
else:
    cuadrante = "CUADRANTE D · Zona Desaconsejada / Alto Riesgo"
    arquetipo = "No implementar con IA"
    supervision = "Proceso manual o software determinista tradicional (SQL / ERP)."
    veredicto = "Riesgo desproporcionado frente al valor generado. Rechazar la iniciativa de IA generativa."

from IPython.display import display, Markdown

reporte = f"""
### Dictamen Directivo de Gobernanza: {nombre_activo}

| Dimensión de Análisis | Diagnóstico Asignado |
|:---|:---|
| **Proceso Evaluado** | {nombre_activo} ({volumen_mensual:,} solicitudes/mes) |
| **Capacidad Principal** | {capacidad_ia} |
| **Cuadrante de la Matriz** | **{cuadrante}** |
| **Arquetipo de Trabajo** | **{arquetipo}** |
| **Mandato de Supervisión** | {supervision} |
| **Recomendación Ejecutiva** | **{veredicto}** |
"""
display(Markdown(reporte))
```

```python
#@title Visualización Gráfica: Matriz de Priorización Valor vs. Costo del Error { run: "auto" }
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8.5, 6.0), dpi=100)

# Fondos de los cuatro cuadrantes
ax.fill_between([0, 5], 5, 10, color='#004F45', alpha=0.12, label='Cuadrante A (Automatización Asistida)')
ax.fill_between([5, 10], 5, 10, color='#C0392B', alpha=0.12, label='Cuadrante B (Co-piloto HITL Obligatorio)')
ax.fill_between([0, 5], 0, 5, color='#7F8C8D', alpha=0.10, label='Cuadrante C (Exploración / Eficiencia)')
ax.fill_between([5, 10], 0, 5, color='#D35400', alpha=0.15, label='Cuadrante D (Zona Desaconsejada)')

# Líneas divisorias de la matriz
ax.axvline(5.0, color='#222222', linestyle='--', linewidth=1.2)
ax.axhline(5.0, color='#222222', linestyle='--', linewidth=1.2)

# Rótulos de cuadrantes
ax.text(2.5, 9.2, 'CUADRANTE A\nAutomatización Asistida\n(Alto Valor · Bajo Riesgo)', ha='center', va='center', fontsize=10, fontweight='bold', color='#004F45')
ax.text(7.5, 9.2, 'CUADRANTE B\nCo-piloto HITL Obligatorio\n(Alto Valor · Alto Riesgo)', ha='center', va='center', fontsize=10, fontweight='bold', color='#A5281B')
ax.text(2.5, 1.2, 'CUADRANTE C\nExploración Interna\n(Bajo Valor · Bajo Riesgo)', ha='center', va='center', fontsize=10, fontweight='bold', color='#555555')
ax.text(7.5, 1.2, 'CUADRANTE D\nZona Desaconsejada\n(Bajo Valor · Alto Riesgo)', ha='center', va='center', fontsize=10, fontweight='bold', color='#B9770E')

# Punto del proceso evaluado
ax.scatter([x_err], [y_val], color='#004F45', s=240, edgecolors='#FFFFFF', linewidth=2.5, zorder=5)
ax.annotate(
    f'{nombre_activo}',
    (x_err, y_val),
    textcoords='offset points',
    xytext=(0, 15),
    ha='center',
    fontsize=11,
    fontweight='bold',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#FFFFFF', edgecolor='#004F45', alpha=0.95),
    zorder=6,
)

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel('Costo del Error e Irreversibilidad del Fallo (Riesgo Directivo) →', fontsize=11, fontweight='bold', color='#222222', labelpad=10)
ax.set_ylabel('Valor de Negocio y Ahorro de Horas (Impacto) →', fontsize=11, fontweight='bold', color='#222222', labelpad=10)
ax.set_title('Ubicación Estratégica en la Matriz Valor vs. Costo del Error', fontsize=13, fontweight='bold', color='#004F45', pad=14)

ax.set_xticks([0, 2.5, 5.0, 7.5, 10])
ax.set_xticklabels(['Inocuo', 'Bajo', 'Umbral de Control', 'Alto', 'Catastrófico'])
ax.set_yticks([0, 2.5, 5.0, 7.5, 10])
ax.set_yticklabels(['Marginal', 'Bajo', 'Umbral de Retorno', 'Medio/Alto', 'Estratégico'])

plt.tight_layout()
plt.show()
```

## 6. Síntesis directiva de la sesión

Para concluir el taller, consolidemos las cinco conclusiones rectoras de la Clase 01:

1. **La IA generativa no reemplaza al ML predictivo:** Trabajan en equipo. El Machine Learning predictivo optimiza decisiones sobre datos tabulares cuantitativos; la IA generativa procesa, extrae y transforma el lenguaje no estructurado.
2. **El valor está en transformar, no en chatear:** Las cuatro capacidades esenciales (resumir, clasificar, extraer estructurado y reescribir) permiten encadenar flujos operativos automatizados hacia ERP y planillas de gestión.
3. **La alucinación es estructural:** Los modelos de lenguaje son motores estadísticos de predicción de tokens, no bases de datos deterministas. Todo diseño de procesos debe asumir que el modelo puede alucinar con alta elocuencia formal.
4. **La privacidad exige contratos Enterprise:** Introducir datos corporativos en herramientas gratuitas expone a la empresa a fuga de secretos comerciales y litigios de propiedad intelectual.
5. **El costo del error define la adopción:** En tareas de alto impacto o falla irreversible, el **Humano en el Bucle (HITL)** bajo una estrategia Centauro es una salvaguarda institucional innegociable.
